```python
# resolve_project_path
# 파일 경로 자동 해결:대 경로를 프로젝트 루트 기준의 절대 경로로 자동 변환하여, 실행 위치(Current Working Directory)가 바뀌더라도 파일 경로가 깨지지 않게 보장하는 것

print("파일 경로 자동 해결")

root = Path.cwd().parent

print(f"root: {root} \n")

print("-----"*10 + "\n")

raw_path = resolve_project_path(root, cfg["data"]["raw_csv"])

raw_path.exists()

print(raw_path)

print("-----"*10 + "\n")

cleaned_path = resolve_project_path(root, cfg["data"]["cleaned_csv"])

cleaned_path.exists()

print(cleaned_path)

print("-----"*10 + "\n")

# SQL lite DB 파일 경로 자동 해결

print("SQL lite DB 파일 경로 자동 해결")

db_path = resolve_project_path(root, cfg["data"]["sqlite_db"])

db_path.exists()

print(db_path)
```
