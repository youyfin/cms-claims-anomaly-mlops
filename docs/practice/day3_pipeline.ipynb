# Supervise + Anomaly + Pipeline


## 1. Supervised
- 관련파일: supervised.py

```python
from pathlib import Path

from cms_claims.supervised import train_high_cost_classifier

root = Path.cwd()

print(root)

while not (root / "config" / "config.yaml").exists() and root != root.parent:
    root = root.parent

print(root)

result = train_high_cost_classifier(
    cleaned_csv=root / "data" / "processed" / "inpatient_claims_clean.csv",
    artifact_dir=root / "artifacts",
    random_state=42,
    test_size=0.25,
    class_weight="balanced",
    n_estimators=300,
    mlflow_cfg=None,
)

result["metrics"]
```

## 2. Anomaly: Unsupervised
- 관련파일: anomaly.py


```python
from pathlib import Path

from cms_claims.anomaly import train_anomaly_model
root = Path.cwd()
print(root)
while not (root / "config" / "config.yaml").exists() and root != root.parent:
    root = root.parent

anomaly_result = train_anomaly_model(
    provider_features_csv=root / "data" / "processed" / "provider_features.csv",
    artifact_dir=root / "artifacts",
    random_state=42,
    contamination=0.05,
    n_estimators=300,
    top_n=20,
    mlflow_cfg=None,
)

anomaly_result["metrics"]
```
